# ESTÁTISTICAS PARA DEVS

## 1. Importações

```python
# Importação das bibliotecas

import pandas as pd
import scipy
```

## 2. Criação do Dataframe

```python
# Principais estruturas do pandas

# Series: É um conjunto de dados unidimensional ou de uma variável só. Como por exemplo, uma coluna de uma planilha.
# DataFrame: É uma estrutura de dados bidimensional, ou seja, é uma tabela, onde cada coluna é uma Series. (MAIS UTILIZADA HOJE EM DIA)

dict_students = {
    'age': [15, 18, 25, 25, 40, 55, 58, 60, 80],
    'height': [160, 162, 165, 168, 172, 174, 174, 174, 176]
}

# Criando um DataFrame
df_students = pd.DataFrame(dict_students)

df_students
```

```python
# Criação de um dataframe de vendas

dict_sales = {'category': ['shirt', 'pants', 'hat', 'socks', 'jacket', 'shoes'],
                'value': [400000, 200000, 150000, 50000, 100000, 300000],
                'quantity': [10000, 5000, 1000, 500, 200, 2000]
    }

df_sales = pd.DataFrame(dict_sales)
df_sales
```

```python
# Criação de um dataframe de dados temporais
dict_invoicing = {
    'date': ['2019-01-01', '2020-01-01', '2021-01-01', '2022-01-01', '2023-01-01'],
    'value': [400000, 800000, 500000, 800000, 900000]
}

df_invoicing = pd.DataFrame(dict_invoicing)
df_invoicing
    
```

```python
df_invoicing.info()

# Convertendo a coluna 'date' para o tipo datetime
df_invoicing['date'] = pd.to_datetime(df_invoicing['date'])

df_invoicing.info()
```

## 3. Medidas

### 3.1 Medidas de Posição

```python
# EM IDADE:
# MÉDIA (Soma de todos os valores dividido pela quantidade de valores)
print(df_students['age'].mean())

# MEDIANA (Valor que separa a metade maior da metade menor de um conjunto de dados)
print(df_students['age'].median())

# MODA (Valor que mais se repete em um conjunto de dados)
print(df_students['age'].mode())

# EM ALTURA:
# MÉDIA (Soma de todos os valores dividido pela quantidade de valores)
print(df_students['height'].mean())

# MEDIANA (Valor que separa a metade maior da metade menor de um conjunto de dados)
print(df_students['height'].median())

# MODA (Valor que mais se repete em um conjunto de dados)
print(df_students['height'].mode())
```

### 3.2 Medidas de Dispersão

```python
# EM IDADE:
# VARIÂNCIA (Média dos quadrados das diferenças entre cada valor e a média aritmética)
print(df_students.age.var())

# DESVIO PADRÃO (Raiz quadrada da variância, é expresso na mesma escala dos dados)
print(df_students.age.std())

# COEFICIENTE DE VARIAÇÃO (Desvio padrão dividido pela média)
print(df_students.age.std() / df_students.age.mean()) # Multiplicado por 100 para obter o valor em porcentagem

# EM ALTURA:
# VARIÂNCIA (Média dos quadrados das diferenças entre cada valor e a média aritmética)
print(df_students.height.var())

# DESVIO PADRÃO (Raiz quadrada da variância, é expresso na mesma escala dos dados)
print(df_students.height.std())

# COEFICIENTE DE VARIAÇÃO (Desvio padrão dividido pela média)
print(df_students.height.std() / df_students.height.mean()) # Multiplicado por 100 para obter o valor em porcent
```

### 3.3 Medidas de Forma

```python
# EM IDADE:
# ASSIMETRIA
print(df_students.age.skew())

# CURTOSE
print(df_students.age.kurtosis())

# EM ALTURA:
# ASSIMETRIA
print(df_students.height.skew())

# CURTOSE
print(df_students.height.kurtosis())
```

```python
# Medidas estatísticas descritivas
print(df_students.describe())
print('\n', + df_students.age.describe())
print('\n', + df_students.height.describe())
```

## 4. Correlação

```python
# PEARSON
# Método de Pearson, sem selecionar explicitamente, usado quando tenho apenas duas variáveis
print(df_students.corr())

# Método de Pearson, selecionando explicitamente, usado quando tenho três ou mais variáveis
print(df_students['age'].corr(df_students['height']))
```

```python
# SPEARMAN
# Método de Spearman, sem selecionar explicitamente, usado quando tenho apenas duas variáveis
print(df_students.corr(method='spearman'))

# Método de Spearman, selecionando explicitamente, usado quando tenho três ou mais variáveis
print(df_students['age'].corr(df_students['height'], method='spearman'))
```

### 5.2 Gráfico de Barras

```python
# Gráfico de barras vertical: vendas (valor)
# O gráfico de barras horizontal é mais utilizado para valores que não ordenados
df_sales.plot.bar(x='category', y='value')
```

```python
# Gráfico de barras vertical: vendas (valor ordenado)
df_sales.sort_values(by='value', ascending=False).plot.bar(x='category', y='value')
```

```python
# Gráfico de barras horizontal: vendas (quantidade)
# O gráfico de barras horizontal é mais utilizado para valores que devem estar ordenados
# Ele começa de maneira inversa
df_sales.plot.barh(x='category', y='quantity')
```

```python
# Gráfico de barras horizontal: vendas (quantidade ordenado)
# Ele começa de maneira inversa, então o "ascending" é True (padrão)
df_sales.sort_values(by='quantity', ascending=True).plot.barh(x='category', y='quantity')
```
